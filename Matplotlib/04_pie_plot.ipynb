```python
import matplotlib.pyplot as plt
import numpy as np

x = [10, 20, 30, 40]
y = ['C', 'C++', 'Java', 'Python']

# plt.pie(x, labels=y)

ex = [0.2, 0.0, 0.2, 0.0]
c = ['g', 'b', 'm', 'y']
plt.pie(x, labels=y, explode=ex, colors=c, autopct='%.2f%%', shadow=True, radius=1, labeldistance=1.2, startangle=90,
textprops={"fontsize":15}, counterclock=False, wedgeprops={"linewidth":1.5, "width":1, 'edgecolor':'m'},
rotatelabels=False)

plt.title("Boss")
plt.legend(loc=2)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = [10, 20, 30, 40]
y = ['C', 'C++', 'Java', 'Python']

plt.pie(x, labels=y,radius=1.5, labeldistance=1.1)
plt.pie([1], colors='w')
plt.title("Donut Chart")

plt.axis(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = [10, 20, 30, 40]
y = ['C', 'C++', 'Java', 'Python']

plt.pie(x, labels=y,radius=1.5)
cr = plt.Circle(xy=(0, 0), radius=1, facecolor='w')
plt.gca().add_artist(cr)
plt.title("Donut Chart")

plt.axis('equal') # For Perfect Circle
plt.show()
```

Pie Chart -> A pie chart shows proportions / percentages of a whole
📌 Total of all values = 100%

Important Parameters of plt.pie()
| Parameter    | Purpose         | Example          |
|  x           | Data values     |  [40,30,20]      |
|  labels      | Slice names     |  ['A','B','C']   |
|  autopct     | Show percentage |  '%1.1f%%'       |
|  colors      | Slice colors    |  ['red','blue']  |
|  explode     | Separate slices |  (0.1,0,0)       |
|  startangle  | Rotation angle  |  90              |
|  shadow      | Shadow effect   |  True            |
|  radius      | Pie size        |  1.2             |
