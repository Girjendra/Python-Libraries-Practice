Axis Functions : In Matplotlib, axis functions are used to control, customize, and format the X-axis and Y-axis of a plot.

1. Axis Labels : Sets label for axes.
➤ plt.xlabel() / ax.set_xlabel()
➤ plt.ylabel() / ax.set_ylabel()
Example : 
plt.xlabel("Time")
plt.ylabel("Time")

2. Axis Limits : Sets minimum and maximum values for axes.
➤ plt.xlim() / ax.set_xlim()
➤ plt.ylim() / ax.set_ylim()
➤ plt.axis([xmin, xmax, ymin, ymax])

Example :
plt.xlim(0, 10)
ax.set_xlim(0, 10)

3. Axis Ticks (IMPORTANT ⭐) : Sets tick positions and labels on axes.
➤ plt.xticks() / ax.set_xticks()
➤ plt.yticks() / ax.set_yticks()

4. Tick Parameters (Formatting)
➤ ax.tick_params()
Controls tick size, color, direction, rotation.

5. Axis Scale : Sets scale of axes.
➤ plt.xscale() / ax.set_xscale()
➤ plt.yscale() / ax.set_yscale()
Example:
plt.xscale('log')
ax.set_xscale('log')

6. Axis Visibility : Controls axis display.
➤ plt.axis()
Example:
plt.axis('on')
plt.axis('off')

7. Grid Control : Adds grid lines.
➤ plt.grid() / ax.grid()
Example:
plt.grid(True)
ax.grid(True)

```python
import matplotlib.pyplot as plt

x = [1,2, 3, 4, 5]
y = [3, 1, 5, 2, 4]

plt.plot(x, y)
lab = ['Python', 'C', 'C++', 'Java', 'HTML']
plt.xticks(x, labels=lab)

plt.show()
```
