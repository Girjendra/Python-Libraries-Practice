Histogram -> A histogram shows the distribution of numerical data by grouping values into bins (intervals).

📌 Histogram ≠ Bar Chart
Histogram → continuous data
Bar chart → categorical data

Important Parameters of plt.hist()
| Parameter   | Purpose             | Example      |
|  x          | Data values         |  [10,20,30]  |
|  bins       | Number of intervals |  bins=5      |
|  color      | Bar color           |  'blue'      |
|  edgecolor  | Border color        |  'black'     |
|  alpha      | Transparency        |  0.7         |
|  label      | Legend label        |  'Scores'    |
|  density    | Probability density |  True        |
