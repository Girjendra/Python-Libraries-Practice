```python
import matplotlib.pyplot as plt

x = [3, 2, 5, 1, 6]
y = [1, 4, 2, 5, 4]

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Basic Stem Plot")

plt.stem(x, y, linefmt=":", markerfmt='ro', bottom=0, basefmt='g', label="Python", orientation='vertical')

plt.legend()
plt.grid(True)
plt.show()
```

Stem Plot -> A stem plot displays discrete data points with:
A vertical line (stem) from baseline
A marker (dot) at the data value
📌 It looks like lollipops

| Parameter             | Purpose          | Example   |
|  x                    | X-axis values    |  [1,2,3]  |
|  y                    | Y-axis values    |  [4,5,6]  |
|  linefmt              | Stem line style  |  'r-'     |
|  markerfmt            | Marker style     |  'bo'     |
|  basefmt              | Baseline style   |  'k-'     |
|  use_line_collection  | Faster rendering |  True     |
