Subplot -> A subplot allows you to display multiple plots in the same figure, arranged in rows and columns.
📌 Each subplot is an independent axes inside one figure.

Basic subplot() Syntax : plt.subplot(nrows, ncols, index)
nrows → number of rows
ncols → number of columns
index → position (1-based)
Positions are filled left to right, top to bottom.

plt.subplot(221) = plt.subplot(2, 2, 1)

plt.subplots() (Recommended Method) ⭐
Syntax : fig, ax = plt.subplots(rows, cols)
fig → entire figure
ax → individual subplot axes(array of plots)

Important Parameters of plt.subplots() : 
| Parameter | Purpose           |
|  nrows    | Number of rows    |
|  ncols    | Number of columns |
|  figsize  | Size of figure    |
|  sharex   | Share x-axis      |
|  sharey   | Share y-axis      |

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]

plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title('Square')


plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title('Line')

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]

fig, ax1 = plt.subplots(1, 2, figsize=(8, 4), sharex=True, sharey=False)

ax1[0].plot(x, y1)
ax1[0].set_title("Square")
ax1[0].set_xlabel("x-axis")
ax1[0].set_ylabel("y-axis")

ax1[1].plot(x, y2)
ax1[1].set_xlabel("x-axis")
ax1[1].set_ylabel("y-axis")
ax1[1].set_title("Line")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]

fig, ax1 = plt.subplots(2, 2)

ax1[0, 0].plot(x, y1)
ax1[0, 0].set_title("Square")

ax1[0, 1].plot(x, y2)
ax1[0, 1].set_title("Line")

ax1[1, 0].plot(x, y2)
ax1[1, 0].set_title("Line")

ax1[1, 1].plot(x, y1)
ax1[1, 1].set_title("Square")

plt.tight_layout()
plt.show()
```

📊 Functions Related to Subplots in Matplotlib1

1️⃣ plt.tight_layout() : Automatically adjusts spacing between subplots to avoid overlapping labels and titles.
Syntax : plt.tight_layout()

2️⃣ fig.subplots_adjust() : Manually adjusts spacing between subplots.
Syntax : fig.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1, wspace=0.4, hspace=0.4)

3️⃣ sharex and sharey : Shares x-axis or y-axis among subplots.
Example : fig, ax = plt.subplots(2, 1, sharex=True)

4️⃣ fig.suptitle() : Adds a common title to all subplots.
Example : fig.suptitle("Overall Figure Title")

5️⃣ ax.set_title() : Sets title for an individual subplot.
Example : ax[0].set_title("Subplot 1")

6️⃣ ax.set_xlabel() and ax.set_ylabel() : Labels x-axis and y-axis of individual subplots.
Example : ax[0].set_ylabel("Subplot 1")

7️⃣ ax.legend() : Adds legend to a subplot.
Example : ax.legend()

8️⃣ ax.grid() : Adds grid lines to a subplot.
Example : ax.grid(True)

9️⃣ plt.delaxes() : Removes a specific subplot.
Example : 
fig, ax = plt.subplots(2, 2)
plt.delaxes(ax[1, 1])

1️⃣1️⃣ ax.set_xlim() and ax.set_ylim() : Sets axis limits for a subplot.

1️⃣2️⃣ ax.text() : Adds text inside a subplot.
Example : ax.text(2, 5, "Important Point")
