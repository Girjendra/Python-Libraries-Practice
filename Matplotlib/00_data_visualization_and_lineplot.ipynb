Data Visualization is the practice of representing data in visual formats—such as charts,
graphs, maps, and dashboards—to make information easier to understand, analyze, and communicate.

Matplotlib is a Python tool that allows you to draw plots, charts, and graphs such as line
graphs, bar charts, pie charts, and histograms.

Common types of plots in Matplotlib
📈 Line plot
📊 Bar chart
🥧 Pie chart
📉 Histogram
🔵 Scatter plot

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 2, 8, 10]

plt.plot(x, y, label='Line 1', marker='s')
plt.plot(x, [1,3,9,7,9], 'ro--', label='Line 2')

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Basic Line Plot")
plt.grid(True)
plt.xlim(1, 5)
plt.ylim(0, 12)

plt.legend()
plt.show()
```

Line Plot -> A line plot displays data points connected by straight line segments.

Line Plot Syntax : plt.plot(x, y)

Important Parameters of plt.plot():
| Parameter    | Purpose        | Example   |
|  x           | X-axis values  |  [1,2,3]  |
|  y           | Y-axis values  |  [4,5,6]  |
|  color       | Line color     |  'blue'   |
|  linewidth   | Line thickness |  2        |
|  linestyle   | Line style     |  '--'     |
|  marker      | Marker type    |  'o'      |
|  markersize  | Marker size    |  8        |
|  label       | Legend label   |  'Sales'  |

Line Styles (Very Important):
| Style  | Meaning         |
|  '-'   | Solid (default) |
|  '--'  | Dashed          |
|  '-.'  | Dash-dot        |
|  ':'   | Dotted          |

Marker Styles (Important):
| Marker | Shape    |
|  'o'   | Circle   |
|  's'   | Square   |
|  '^'   | Triangle |
|  '*'   | Star     |
|  '+'   | Plus     |

Formatting Shortcut (Important) : Matplotlib allows compact formatting:
plt.plot(x, y, 'ro--')
Meaning:
r → red
o → circle marker
-- → dashed line
