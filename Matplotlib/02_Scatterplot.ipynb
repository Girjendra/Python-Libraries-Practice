```python
from matplotlib import pyplot as plt
import numpy as np

day = [1, 2, 3, 4, 5, 6, 7]
No = [3, 2, 4, 3, 5, 1, 4]
No2 = [3, 7, 1, 6, 3, 2, 5]

plt.title("Scatter plot", fontsize=15, color='g')
plt.xlabel("Day", fontsize=15, color='b')
plt.ylabel("No", fontsize=15, color='m')

size = [200, 150, 329, 500, 250, 200, 400]

color = [23, 45, 43, 41, 59, 40, 24]
plt.scatter(day, No, c='m', s=size)
plt.scatter(day, No2, c=color, s=size, cmap="viridis")
t = plt.colorbar()
t.set_label("ColorBar", fontsize=20)

plt.grid()
plt.show()
```

What is a Scatter Plot -> A scatter plot displays individual data points using dots on a 2D plane.

Syntax : plt.scatter(x, y)

Important Parameters of plt.scatter()
| Parameter    | Purpose         | Example         |
|  x           | X-axis data     |  [1,2,3]        |
|  y           | Y-axis data     |  [4,5,6]        |
|  s           | Marker size     |  s=100          |
|  c           | Color of points |  c='red'        |
|  marker      | Marker style    |  'o', '*'       |
|  alpha       | Transparency    |  0.7            |
|  label       | Legend label    |  'Data Points'  |
|  edgecolors  | Border color    |  'black'        |
|  linewidths  | Border width    |  1              |

Marker Styles (Important)
| Marker | Meaning |
|  'o'   | Circle  |
|  '.'   | Point   |
|  '*'   | Star    |
|  '+'   | Plus    |
|  'x'   | Cross   |
|  's'   | Square  |
|  'D'   | Diamond |
