```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 3, 4, 5, 6]


plt.step(x, y, color='r', marker='p', ms=10, mfc='r', label="Series 1", where='post')
plt.step(x, [3, 4, 5, 6, 7], color='g', marker='o', ms=10, mfc='g', label="Series 2", where='pre')

plt.legend(loc=2)
plt.grid()
plt.title("Basic Step Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.show()
```

Step Plot -> A step plot displays data as horizontal and vertical line segments, forming a staircase-like shape.
📌 Unlike line plots, values change suddenly at certain points.

Step Plot Syntax : plt.step(x, y)
Important Parameters of plt.step()
| Parameter   | Purpose        | Example                    |
|  x          | X-axis values  |  [1,2,3]                   |
|  y          | Y-axis values  |  [4,5,6]                   |
|  where      | Step position  |  'pre', 'post', 'mid'      |
|  color      | Line color     |  'blue'                    |
|  linewidth  | Line thickness |  2                         |
|  linestyle  | Line style     |  '--'                      |
|  label      | Legend label   |  'Data'                    |

where Parameter (VERY IMPORTANT): Controls where the step occurs.
'pre' : Step occurs before the x-value.
'post' (default) : Step occurs after the x-value.
'mid' : Step occurs at the midpoint.
