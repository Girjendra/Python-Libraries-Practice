plt.savefig() : plt.savefig() is a Matplotlib function used to save a plot or graph as an image file on the computer.

Basic Syntax : plt.savefig("filename.ext")
"filename" → name of the file
"ext" → file format (png, jpg, pdf, etc.)

Common File Formats Supported : 
| Format | Extension | Use                    |
| PNG    |  .png     | Best for general use   |
| JPEG   |  .jpg     | Smaller size images    |
| PDF    |  .pdf     | High quality documents |
| SVG    |  .svg     | Vector graphics        |
| EPS    |  .eps     | Publishing             |

⭐ Most Commonly Used savefig() Parameters : 
|   Parameter   |   Purpose                  |   Common Example          |
|  fname        | File name / path           |  "graph.png"              |
|  dpi          | Image quality (resolution) |  dpi=300                  |
|  format       | File format                |  format='png'             |
|  bbox_inches  | Removes extra white space  |  bbox_inches='tight'      |
|  transparent  | Transparent background     |  transparent=True         |
|  pad_inches   | Padding around figure      |  pad_inches=0.1           |
|  facecolor    | Background color           |  facecolor='white'        |
|  edgecolor    | Border color               |  edgecolor='black'        |
|  orientation  | Page orientation (PDF)     |  orientation='landscape'  |


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 50,30, 60]

plt.plot(x, y, color='g')
plt.title("Simple Line Graph")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.savefig("line_graph.png", dpi=1000, bbox_inches='tight', transparent=False, facecolor='g', edgecolor='g')
plt.show()
```
