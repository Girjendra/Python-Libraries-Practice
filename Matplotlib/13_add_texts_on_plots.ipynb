📝 Adding Simple Text Inside a Plot
syntax : plt.text(x, y, "Text")
📌 Places text at coordinates (x, y).

Important Parameters of plt.text() : 
| Parameter    | Description          | Example    |
|  x           | X-coordinate         |  2         |
|  y           | Y-coordinate         |  20        |
|  s           | Text string          |  "Hello"   |
|  fontsize    | Text size            |  12        |
|  color       | Text color           |  'red'     |
|  fontweight  | Font weight          |  'bold'    |
|  ha          | Horizontal alignment |  'center'  |
|  va          | Vertical alignment   |  'top'     |

Alignment Options : 
| Parameter | Values                        |
|  ha       |  'left', 'center',  'right'   |
|  va       |  'top', 'center',  'bottom'   |

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 4]

plt.plot(x, y)
plt.text(2, 4, "Text", fontsize=20, bbox={'facecolor':'r'}, ha='center', va='bottom', rotation=45)

plt.show()
```

plt.annotate() : Is used to add text with an optional arrow pointing to a specific data point in a plot.

Syntax : plt.annotate(text, xy, xytext=None, arrowprops=None)
Meaning of Main Parameters : 
| Parameter    | Description           |
|  text        | Text to display       |
|  xy          | Point being annotated |
|  xytext      | Position of text      |
|  arrowprops  | Arrow properties      |

Important Arrow Properties (arrowprops) : 
| Property     | Meaning           | Example                 |
|  arrowstyle  | Arrow shape       |  "->", "-", "<-", "<->" |
|  color       | Arrow color       |  'red'                  |
|  lw          | Line width        |  2                      |
|  shrink      | Shrink arrow ends |  0.05                   |

Text Formatting Parameters : 
| Parameter    | Purpose        | Example  |
|  fontsize    | Text size      |  12      |
|  color       | Text color     |  'blue'  |
|  fontweight  | Text thickness |  'bold'  |
|  rotation    | Text rotation  |  30      |



```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 4]

plt.plot(x, y)
plt.annotate("Peak",xy=(3, 5), xytext=(2, 4), arrowprops=dict(arrowstyle="<->", color='r'), color='g', bbox={'facecolor':'r'},
            ha='center', va='top', rotation=45 )

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [3, 1, 5, 2, 4]

plt.plot(x, y)
plt.annotate("Peak",
             xy=(3, 5), xytext=(2, 4), arrowprops=dict(arrowstyle="<->", color='r'), color='g', bbox={'facecolor':'r'},
             xycoords='data', textcoords='offset points')

plt.show()
```

xycoords : It defines the coordinate system of the point being annotated (xy).
Default:
xycoords='data'
Means:The point (x, y) is interpreted in data units

Common xycoords Values : 
| xycoords  Value     | Meaning                           |
|  'data'             | Uses actual data values (default) |
|  'axes fraction'    | Relative to axes (0 to 1)         |
|  'figure fraction'  | Relative to entire figure         |
|  'offset points'    | Offset in points                  |

textcoords : It defines how the text position (xytext) is interpreted.
📌 Default: textcoords = xycoords
