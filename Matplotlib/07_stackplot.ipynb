```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
area1 = [2, 3, 2, 5, 4]
area2 = [3, 2, 4, 1, 5]
area3 = [4, 3, 5, 3, 2]
l = ["area1", "area2", "area3"]
c = ['r', 'g', 'm']
plt.stackplot(x, area1, area2, area3, labels=l, colors=c, baseline='weighted_wiggle')

plt.xlabel("Time")
plt.ylabel("Values")
plt.title("Basic Stack Plot")
plt.legend()
plt.legend(loc=2)
plt.show()
```

Stack Plot -> A stack plot shows multiple data series stacked on top of each other over a common x-axis.

Think of it as multiple area plots stacked together.
Stack Plot Syntax : plt.stackplot(x, y1, y2, ..., labels=[])

Important Parameters of plt.stackplot()
| Parameter   | Purpose         | Example          |
|  x          | X-axis values   |  [1,2,3]         |
|  y1, y2...  | Data series     |  [2,3,4]         |
|  labels     | Legend labels   |  ['A','B']       |
|  colors     | Area colors     |  ['red','blue']  |
|  alpha      | Transparency    |  0.7             |
|  baseline   | Stack reference |  'zero'          |

baseline Options in plt.stackplot()
| Baseline Value      | Description                               | Use Case               |
|  'zero'             | Stacks areas starting from zero (default) | Normal cumulative data |
|  'sym'              | Centers stack around zero                 | Stream graph style     |
|  'wiggle'           | Minimizes wiggle between layers           | Trend visualization    |
|  'weighted_wiggle'  | Improved wiggle minimization              | Smooth stream graphs   |
