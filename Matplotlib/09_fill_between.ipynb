```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([1, 2, 3, 4, 5])
y1 = [2, 4, 6, 5, 7]
y2 = [1, 3, 4, 3, 5]

# plt.fill_between(x, y, color='green', alpha=0.5, where= (x >= 2) & (x <= 4))
plt.fill_between(x, y1, y2, color='green', alpha=0.5)

plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("Simple linear plot")
plt.show()
```

fill_between() - > fill_between() is used to shade (fill) the area between two curves or between a curve and a baseline.

Basic Syntax : plt.fill_between(x, y1, y2=0)
x → x-axis values
y1 → upper curve
y2 → lower curve (default = 0)

Important Parameters of fill_between() : 
| Parameter     | Purpose             | Example   |
|  x            | X-axis values       |  [1,2,3]  |
|  y1           | Upper boundary      |  [3,4,5]  |
|  y2           | Lower boundary      |  0        |
|  where        | Conditional filling |  y > 3    |
|  color        | Fill color          |  'blue'   |
|  alpha        | Transparency        |  0.5      |
|  interpolate  | Smooth boundary     |  True     |
|  label        | Legend label        |  'Area'   |
