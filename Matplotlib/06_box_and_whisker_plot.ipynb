```python
import matplotlib.pyplot as plt 

x = [10, 20, 30, 40, 50, 60, 70, 120]

plt.boxplot(x, notch=False, vert=True, widths=0.2, tick_labels=["Boss"], label="Python",
patch_artist=False, showmeans=True, whis=1, sym='g+', boxprops=dict(color='g'), capprops=dict(color='r'),
whiskerprops=dict(color='m'), flierprops=dict(markeredgecolor='y'))

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt 

x = [10, 20, 30, 40, 50, 60, 70, 120]
x1 = [10, 20, 40, 50, 30, 60, 90, 80]

y = [x, x1]
plt.boxplot(y, notch=False, vert=True, widths=0.2, tick_labels=["Python", 'C++'], label="Box plot",
patch_artist=False, showmeans=True, whis=1, sym='g+', boxprops=dict(color='g'), capprops=dict(color='r'),
whiskerprops=dict(color='m'), flierprops=dict(markeredgecolor='y'))

plt.legend()
plt.show()
```

Box and Whisker Plot-> A Box and Whisker Plot is a statistical graph used to display the distribution of numerical data using five key summary values:
Minimum
First Quartile (Q1)
Median (Q2)
Third Quartile (Q3)
Maximum

It helps in understanding spread, central tendency, variability, and outliers.

Why it is called “Box and Whisker”
The box represents the interquartile range (IQR) (Q1 to Q3)
The line inside the box represents the median
The whiskers extend from the box to show the minimum and maximum values

🔹 Five-Number Summary:
| Term        | Meaning                             |
| Minimum     | Smallest value (excluding outliers) |
| Q1          | 25% of data below this value        |
| Median (Q2) | Middle value of data                |
| Q3          | 75% of data below this value        |
| Maximum     | Largest value (excluding outliers)  |

Components of a Box Plot
1. Box
🔹 Represents the range from Q1 to Q3
🔹 Length of box = Interquartile Range (IQR)
2. Median Line
🔹 A line inside the box
🔹 Shows the central value
3. Whiskers
🔹 Extend from Q1 to minimum and Q3 to maximum
🔹 Show the spread of data
4. Outliers
🔹 Data points outside the whiskers
🔹 Shown as dots or circles

IQR=Q3−Q1

Outlier rule:
Below: Q1−1.5×IQR
Above:Q3+1.5×IQR

```python
data = [
    [45, 50, 55, 60, 65],
    [55, 60, 65, 70, 75],
    [65, 70, 75, 80, 85]
]

plt.boxplot(data, tick_labels=["Class A", "Class B", "Class C"])
plt.title("Comparison Using Box Plot")
plt.show()
```
