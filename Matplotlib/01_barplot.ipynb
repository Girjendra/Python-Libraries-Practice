```python
from matplotlib import pyplot as plt
import numpy as np

x = ["python", "C", "C++", "Java"]
y = [85, 70, 60, 82]
c = ["m", "g", "r", "b"]

plt.xlabel("Language", fontsize=20, color="r")
plt.ylabel("Percentage", fontsize=20, color="r")
plt.title('Bar Plot Example')

plt.bar(x, y, color=c, width=.5, label="Popularity")
plt.legend(fontsize=10)
plt.show()
```

📊 What is a Bar Plot -> A bar plot displays categorical data on one axis and numerical values on the other axis.
The height (or length) of each bar represents the value of that category.

📊 Types of bar plots
Vertical bar plot → plt.bar()
Horizontal bar plot → plt.barh()
Grouped bar plot
Stacked bar plot

1️⃣ Vertical Bar Plot -> A vertical bar plot displays data using vertical bars,
    where the height of each bar represents the value of a category.
👉 Created using plt.bar().

2️⃣ Horizontal Bar Plot -> A horizontal bar plot represents data using horizontal bars,
    where the length of each bar shows the value of a category.
👉 Created using plt.barh().

3️⃣ Grouped Bar Plot -> A grouped bar plot displays multiple bars side-by-side for each category,
    allowing comparison of different groups within the same category.

4️⃣ Stacked Bar Plot -> A stacked bar plot shows multiple data values stacked on top of each other within a single bar,
    representing both individual contributions and total value.

```python
from matplotlib import pyplot as plt
import numpy as np

x = ["python", "C", "C++", "Java"]
y = [85, 70, 60, 82]
z = [40, 50, 50, 56]

plt.xlabel("Language", fontsize=20, color="r")
plt.ylabel("Percentage", fontsize=20, color="r")
plt.title('Stacked Bar Plot Example')
width=.4
p = np.arange(len(x))
p1 = [j+width for j in p]

plt.bar(x, y,  width, color="r", label="Popularity1")
plt.bar(x, z,  width, color="y", label="Popularity2")

plt.xticks(p+width/2,x, rotation=90)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

subjects = ['Math', 'Science', 'English']
boys_marks = [75, 80, 70]
girls_marks = [85, 78, 88]

x = np.arange(len(subjects))   # label locations
width = 0.35                   # width of bars

plt.bar(x - width/2, boys_marks, width, label='Boys')
plt.bar(x + width/2, girls_marks, width, label='Girls')

plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Grouped Bar Plot Example')
plt.xticks(x, subjects)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

years = ['2022', '2023', '2024']
product_A = [20, 25, 30]
product_B = [15, 20, 25]

x = np.arange(len(years))

plt.bar(x, product_A, label='Product A')
plt.bar(x, product_B, bottom=product_A, label='Product B')

plt.xlabel('Years')
plt.ylabel('Sales')
plt.title('Stacked Bar Plot Example')
plt.xticks(x, years)
plt.legend()

plt.show()
```
