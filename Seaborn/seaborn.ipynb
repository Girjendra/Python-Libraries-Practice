```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
data = sns.load_dataset('penguins')
```

```python
print(data.head(5))
```

```python
print(data.columns)
print(data['species'].value_counts())
print(data['island'].value_counts())
```

```python
# sns.set_style('whitegrid')
# sns.set_style('dark')
sns.set_style('ticks')

# sns.set_context('talk')
# sns.set_context('poster')
sns.set_context('notebook')


sns.scatterplot(data=data, x = 'bill_length_mm', y='body_mass_g', hue='sex', palette='Dark2')

# sns.despine(left=True)
sns.despine()
```

```python
sns.scatterplot(data=data, x='species', y='body_mass_g', hue='island', style='sex', alpha=0.5)
```

```python
sns.stripplot(data=data, x='species', y='body_mass_g', hue='sex', dodge=True, jitter=True)
# sns.stripplot(data=data, x='species', y='body_mass_g', hue='sex', dodge=True)
# sns.stripplot(data=data, x='species', y='body_mass_g', hue='sex', jitter=True)
```

```python
sns.set_context('talk')
sns.swarmplot(data=data, x='species', y='body_mass_g', hue='island', dodge=True)
```

```python
sns.set_context('notebook')
sns.histplot(data=data, x='body_mass_g', hue='sex', multiple='stack')
```

```python
sns.regplot(data=data, x='body_mass_g', y='flipper_length_mm', scatter=True, color='green')
```

```python
sns.set_style('whitegrid')
sns.lineplot(data=data, x='body_mass_g', y='flipper_length_mm', hue='sex', style='island')
```

```python
sns.jointplot(data=data, x='body_mass_g', y='flipper_length_mm', hue='island', kind='scatter')
```

```python
sns.barplot(data=data, x='species', y='flipper_length_mm', hue='sex', palette=['pink', 'skyblue'], estimator=np.var)
```

```python
sns.countplot(data=data, x='species', hue='island', palette='Set2')
```

```python
sns.boxenplot(data=data, x='species', y='body_mass_g', hue='sex')
```

```python
# sns.violinplot(data=data, x='species', y='body_mass_g', hue='sex', split=True)
# sns.violinplot(data=data, x='species', y='body_mass_g', hue='sex')

sns.violinplot(data=data, x='species', y='body_mass_g', palette='pastel')
sns.swarmplot(data=data, x='species', y='body_mass_g', hue='island', size=3)
```

```python
sns.kdeplot(data=data, x='body_mass_g', hue='species', fill=True)
```

```python
col = ["bill_length_mm",  "bill_depth_mm",  "flipper_length_mm", "body_mass_g"]
sns.heatmap(data=data[col].corr(), annot=True, cmap='Blues', vmin=.2, linewidths=2, linecolor='black')
```

```python
sns.rugplot(data=data, x='body_mass_g', hue='species', palette='Set2', height=0.2)
```

```python
sns.pairplot(data=data, hue='sex', palette='Set2', diag_kind='kde')
```

```python
graph = sns.PairGrid(data=data, hue='island', palette='Set2')
graph.map_upper(sns.scatterplot)
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot)
```
