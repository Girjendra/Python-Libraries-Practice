## 1: Data Handling in sklearn

### 1.1 Built-in Datasets

```python
import sklearn.datasets as ds
```

```python
print(dir(ds))

for dataset in dir(ds):
    if 'load' in dataset:
        print(dataset)
```

```python
data = ds.load_iris(return_X_y=False, as_frame=False)
```

Parameters:
* return_X_y (bool, default=False)

    → True: returns (X, y) directly

    → False: returns Bunch object

* as_frame (bool, default=False)

    → True: pandas DataFrame

    → False: numpy array

Output (Bunch object contains):
* data → features
* target → labels
* feature_names
* target_names
* DESCR → dataset description

```python
# Only features and target
X, Y = ds.load_iris(return_X_y=True, as_frame=True)
print(X.head())
```

```python
# Bunch Object
data = ds.load_iris(as_frame=True)
X = data.data
Y = data.target
print(X.head())
print(Y.head())
```

### 1.3 Train-Test Split

```python
from sklearn.model_selection import train_test_split
```

X_train, X_test, y_train, y_test = train_test_split(

    X, Y,
    test_size=None,
    train_size=None,
    random_state=None,
    shuffle=True,
)

#### Parameters (IMPORTANT):
1.3.1 test_size
* float → percentage (0.2 = 20%)
* int → exact number of samples

1.3.2 train_size
* alternative to test_size

1.3.3 random_state
* controls randomness
* same value → same split every time

1.3.4 shuffle
* True → shuffle before split
* False → no shuffle

```python
data = ds.load_diabetes()
X = data.data
Y = data.target
```

```python
print(X.shape, Y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=42
)
```
