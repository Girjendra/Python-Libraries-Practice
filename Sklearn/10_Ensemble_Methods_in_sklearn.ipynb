## **10: Ensemble Methods in sklearn**

Ensemble = combining multiple models to make one strong model

### **10.1 Types of Ensemble**

**1. Bagging (Parallel)**
* Many models trained independently
* eg. Random Forest
```bash
from sklearn.ensemble import RandomForestClassifier 
```

**2. Boosting (Sequential)**
* Models learn one after another
* eg. AdaBoost, Gradient Boosting
```bash
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
```

**Voting**
* Combine predictions of different models
* eg. VotingClassifier
```bash
from sklearn.ensemble import VotingClassifier
```

**When to use**
* Random Forest → general use
* Boosting → high accuracy
* Voting → combine best models
