### **4.1 Logistic Regression**

**Syntax:**
```bash
model = LogisticRegression(
    penalty='l2',
    C=1.0,
    fit_intercept=True,
    solver='lbfgs',
    max_iter=100,
    multi_class='auto'
)
```

**Parameters:**
* penalty → 'l1', 'l2', 'elasticnet', None
    * Type of regularization used (to prevent overfitting)
* C → inverse of regularization
    * small C = strong regularization (simpler model)
    * large C = weak regularization (more flexible model) ⭐
* solver → 'lbfgs', 'liblinear', 'saga'
    * Algorithm used to train the model
    * 'lbfgs' is default and works well in most cases
* max_iter → iterations
    * Maximum number of steps to train the model
* multi_class → 'auto', 'ovr', 'multinomial'
    * How to handle multiple classes
    * 'ovr' = one vs rest
    * 'multinomial' = better for multi-class problems

**Methods:**
```bash
model.fit(X_train, y_train)
model.predict(X_test)
model.predict_proba(X_test)
model.score(X_test, y_test)
```

**predict_proba()** returns the probability of each class for a given input.
```bash
model.predict_proba(X_test)
```


````python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Small dataset (hours studied → pass/fail)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])   # 0 = fail, 1 = pass

# Create model
model = LogisticRegression()

# Train model
model.fit(X, y)

# Predict
y_pred = model.predict(X)

print("Predictions:", y_pred)
````

### **4.2 K-Nearest Neighbors (KNN)**

**Syntax**
```bash
model = KNeighborsClassifier(
    n_neighbors=5,
    weights='uniform',
    algorithm='auto',
    metric='minkowski',
    p=2
)
```

**Parameters (simple language)**
* n_neighbors → number of nearest points to consider ⭐
* weights → 'uniform', 'distance'
    * 'uniform' → all neighbors equal
    * 'distance' → closer points get more importance
* algorithm → 'auto', 'ball_tree', 'kd_tree', 'brute'
    * method used to find neighbors (usually keep 'auto')
* metric → distance type used
    * 'minkowski' is default
* p → type of distance
    * 1 = Manhattan distance
    * 2 = Euclidean distance ⭐

p is used only when metric = "minkowski"

**Methods:**
```bash
model.fit(X_train, y_train)
model.predict(X_test)
model.score(X_test, y_test)
```

````python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

# Small dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Create model
model = KNeighborsClassifier(n_neighbors=3)

# Train
model.fit(X, y)

# Predict
y_pred = model.predict(X)

print("Predictions:", y_pred)
````

### **4.3 Support Vector Machine (SVM)**

#### **SVM Classifier (SVC)**
Syntax
```bash
model = SVC(
    C=1.0,
    kernel='rbf',
    degree=3,
    gamma='scale',
    probability=False
)
```

**Parameters (simple language)**
* C → regularization ⭐
    * high C = strict (less error allowed)
    * low C = flexible (more error allowed)
* kernel → 'linear', 'rbf', 'poly', 'sigmoid' ⭐
    * defines shape of decision boundary
* degree → for polynomial kernel
    * used only when kernel='poly'
* gamma → influence of points ⭐
    * high gamma = complex boundary
    * low gamma = smooth boundary
* probability → True/False
    * True → enables probability output (predict_proba)

````python
import numpy as np
from sklearn.svm import SVC

# Small dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Create model
model = SVC(kernel='linear')

# Train
model.fit(X, y)

# Predict
y_pred = model.predict(X)

print("Predictions:", y_pred)
````

#### **SVM Regressor (SVR)**

**Syntax**
```bash
model = SVR(
    kernel='rbf',
    C=1.0,
    epsilon=0.1,
    degree=3,
    gamma='scale'
)
```

**Parameters (simple language)**
* kernel → 'linear', 'rbf', 'poly', 'sigmoid' ⭐
    * decides shape of regression (line or curve)
* C → regularization ⭐
    * high C = strict (fit data closely)
    * low C = flexible (allows more error)
* epsilon (ε) ⭐
    * allowed error margin (small errors ignored)
* degree
    * used only for polynomial kernel
* gamma
    * controls influence of data points
    * high = complex curve, low = smooth curve

**Methods:**
```bash
model.fit(X_train, y_train)
model.predict(X_test)
model.score(X_test, y_test)
```

````python
import numpy as np
from sklearn.svm import SVR

# Small dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

# Create model
model = SVR(kernel='linear')

# Train
model.fit(X, y)

# Predict
y_pred = model.predict(X)
print("Predictions:", y_pred)
````

### **4.4 Decision Tree**

#### **Decision Tree Classifier**

**Syntax**
```bash
model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1
)
```

**Parameters (simple language)**
* criterion → 'gini', 'entropy' ⭐
    * method to measure split quality
    * 'gini' = faster, 'entropy' = more accurate sometimes
* max_depth → depth of tree ⭐
    * controls how deep tree grows
    * smaller value = less overfitting
* min_samples_split
    * minimum samples needed to split a node
* min_samples_leaf
    * minimum samples required at a leaf node

````python
import numpy as np
from sklearn.tree import DecisionTreeClassifier

# Small dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Create model
model = DecisionTreeClassifier(max_depth=2)

# Train
model.fit(X, y)

# Predict
y_pred = model.predict(X)
print("Predictions:", y_pred)
````

    * smaller = less overfitting
* min_samples_split

    * minimum samples required at a leaf node

**Methods:**
```bash
model.fit(X_train, y_train)
model.predict(X_test)

```

````python
import numpy as np
from sklearn.tree import DecisionTreeRegressor

# Small dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])   # continuous output

# Create model
model = DecisionTreeRegressor(max_depth=2)

# Train
model.fit(X, y)

# Predict
y_pred = model.predict(X)
print("Predictions:", y_pred)
````

### **4.5 Random Forest**

#### **Random Forest (Classifier)**

**Syntax**
```bash
model = RandomForestClassifier(
    n_estimators=100,
    criterion='gini',
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)
```

**Parameters (simple language)**
* n_estimators ⭐
    * number of trees in the forest
    * more trees = better accuracy (but slower)
* criterion → 'gini', 'entropy'
    * how to measure split quality
* max_depth ⭐
    * maximum depth of each tree
    * smaller = less overfitting
* min_samples_split
    * minimum samples needed to split a node
* min_samples_leaf
    * minimum samples required at a leaf
* random_state
    * for reproducibility (same result every run)

**For Regressor version just change :**
```bash
from sklearn.ensemble import RandomForestRegressor
```

````python
import numpy as np
from sklearn.ensemble import RandomForestClassifier

# Small dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Create model
model = RandomForestClassifier(n_estimators=10)

# Train
model.fit(X, y)

# Predict
y_pred = model.predict(X)
print("Predictions:", y_pred)
````

### **4.6 Naive Bayes**

**Syntax:**
```bash
model = GaussianNB(priors=None, 
      var_smoothing=1e-9
)
```

**Parameters (simple language)**
* priors → class probabilities
* var_smoothing ⭐
    * small value added to variance
    * prevents division by zero / numerical issues

**Types in sklearn (important)**
1. GaussianNB → for continuous data
2. MultinomialNB → for count data (te2t, word counts)
3. BernoulliNB → for binary data (0/13)

````python
import numpy as np
from sklearn.naive_bayes import GaussianNB

# Small dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Create model
model = GaussianNB()

# Train
model.fit(X, y)

# Predict
y_pred = model.predict(X)
print("Predictions:", y_pred)
````

### **4.7 Evaluation Functions**

**Accuracy:**
```bash
accuracy_score(y_test, y_pred)
```

Confusion Matrix:
```bash
confusion_matrix(y_test, y_pred)
```

Classification Report:
```bash
classification_report(y_test, y_pred)
```

````python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Dataset
X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

# Model
model = LogisticRegression()
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

# Metrics
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
````
