## **8: Model Selection in sklearn**

Goal = Find best model + best parameters

### **8.1 GridSearchCV**

GridSearchCV tries all parameter combinations and finds the best one

**Syntax:**
```bash
model = GridSearchCV(
    estimator=your_model,
    param_grid={
        'param1': [values],
        'param2': [values]
    },
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)
```

**⭐⭐GridSearchCV works with one model at a time⭐⭐**

**Parameters (simple language)**
* estimator ⭐
    * model you want to tune (e.g., SVC, KNN)
* param_grid ⭐
    * dictionary of parameters and values to try
* cv
    * number of cross-validation folds
* scoring
    * metric to evaluate model (e.g., accuracy, f1)
* n_jobs
    * number of CPU cores (-1 = use all)
* verbose (optional)
    * shows progress while training


````python
import numpy as np
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

# Dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Model
model = SVC()

# Parameters to try
param_grid = {
    'C': [0.1, 1, 10],
    'kernel': ['linear', 'rbf']
}

# GridSearch
grid = GridSearchCV(model, param_grid, cv=3)
grid.fit(X, y)

print("Best Params:", grid.best_params_)
print("Best Score:", grid.best_score_)
print(grid.best_estimator_)
````

### **8.2 RandomizedSearchCV**

RandomizedSearchCV tries random combinations of parameters (faster than GridSearchCV)

**Syntax**
```bash
model = RandomizedSearchCV(
    estimator=your_model,
    param_distributions={
        'param1': [values],
        'param2': [values]
    },
    n_iter=10,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    random_state=42
)
```

**Parameters (simple language)**
* estimator ⭐
    * model you want to tune (e.g., SVC, KNN)
* param_distributions ⭐
    * parameters and values to try (randomly sampled)
* n_iter ⭐
    * number of random combinations to try
* cv
    * number of cross-validation folds
* scoring
    * metric to evaluate model (accuracy, f1, etc.)
* n_jobs
    * CPU cores (-1 = use all)
* random_state
    * ensures same random results every time

````python
import numpy as np
from sklearn.model_selection import RandomizedSearchCV
from sklearn.svm import SVC

# Dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

# Model
model = SVC()

# Parameter distributions
param_dist = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf'],
    'gamma': ['scale', 'auto']
}

# Randomized Search
rand = RandomizedSearchCV(model, param_dist, n_iter=5, cv=3, random_state=42)
rand.fit(X, y)

print("Best Params:", rand.best_params_)
print("Best Score:", rand.best_score_)
print(grid.best_estimator_)
````

**Common Parameters to Tune**
