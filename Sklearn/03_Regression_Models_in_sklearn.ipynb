## **3: Regression Models in sklearn**

### **3.1 Linear Regression**

**Syntax:**
```bash
model = LinearRegression(
    fit_intercept=True,
    copy_X=True,
    n_jobs=None,
    positive=False
)
```

**Parameters:**
* fit_intercept -> Whether to calculate intercept/bias b (in y = mx + b) 
* copy_X -> Whether to copy original data or not
* n_jobs -> Number of CPU cores to use
* positive -> Force coefficients to be positive

````python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])   # input (independent variable)
y = np.array([2, 4, 6, 8, 10])            # output (dependent variable)

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print("Predictions:", y_pred)
````

### **3.2 Ridge Regression**

**Syntax:**
```bash
model = Ridge(
    alpha=1.0,
    fit_intercept=True,
    copy_X=True,
    positive=False
)
```

**Parameters:**
* alpha → Regularization strength (controls overfitting, higher = less overfitting)
* fit_intercept → Whether to calculate intercept (bias term)
* copy_X → Copy data or modify original dataset
* positive → Forces coefficients to be positive only

````python
import numpy as np
from sklearn.linear_model import Ridge

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = Ridge(alpha=1.0)
model.fit(X, y)

y_pred = model.predict(X)
print("Ridge Predictions:", y_pred)
````

### **3.3 Lasso Regression**

**Syntax:**
```bash
model = Lasso(
    alpha=1.0,
    fit_intercept=True,
    copy_X=True,
    selection='cyclic',
    positive=False
)
```
**Parameters:**
* alpha → Controls regularization strength (higher = more feature shrinkage) ⭐
* fit_intercept → Whether to calculate intercept (bias term)
* copy_X → Copy data or modify original dataset
* selection → Method to update coefficients
    * 'cyclic' → update one by one
    * 'random' → random update (faster sometimes)
* positive → Forces coefficients to be positive only

````python
import numpy as np
from sklearn.linear_model import Lasso

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = Lasso(alpha=.1)
model.fit(X, y)

y_pred = model.predict(X)
print("Lasso Predictions:", y_pred)
````

### **3.4 Elastic Net**

**Syntax:**
```bash
model = ElasticNet(
    alpha=1.0,
    l1_ratio=0.5,
    fit_intercept=True,
    copy_X=True,
    selection='cyclic'
)
```

**Parameters**
* alpha → Overall regularization strength (higher = more shrinkage) ⭐
* l1_ratio → Mix of Lasso and Ridge
    * 1 → pure Lasso
    * 0 → pure Ridge
    * 0.5 → mix of both
* fit_intercept → Whether to calculate intercept (bias term)
* copy_X → Copy data or modify original dataset
* selection →
    * 'cyclic' → update one by one
    * 'random' → random updates

````python
import numpy as np
from sklearn.linear_model import ElasticNet

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = ElasticNet(alpha=0.1, l1_ratio=0.5)
model.fit(X, y)

y_pred = model.predict(X)
print("ElasticNet Predictions:", y_pred)
````

### **3.5 Multiple Linear Regression**

Same as LinearRegression (just multiple features)

````python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [4, 5]
])   # multiple features

y = np.array([3, 5, 7, 9])

model = LinearRegression()
model.fit(X, y)

print(model.predict(X))
````

### **3.6 Polynomial Regression**

**sklearn does NOT have a direct "PolynomialRegression" model.**

**Instead, we do it in 2 steps:**
1. Convert features → polynomial features
2. Apply Linear Regression

**Syntax:**
```bash
poly = PolynomialFeatures(
    degree=2,
    interaction_only=False,
    include_bias=True
)
```


````python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1, 4, 9, 16, 25])   # y = x^2

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_poly)

print(y_pred)
````

### **3.7 Support Vector Regression(SVR)**

**Syntax:**
```bash
model = SVR(
    kernel='rbf',
    C=1.0,
    epsilon=0.1,
    degree=3,
    gamma='scale'
)
```

**Parameters**
* kernel → Type of function used
    * ('linear', 'poly', 'rbf', 'sigmoid') ⭐
* C → Regularization parameter
    * (high = strict, low = flexible) ⭐
* epsilon (ε) → Error tolerance (margin size) ⭐
    * → ignores small errors
* degree → Used in polynomial kernel
* gamma → Influence of data points
    * (high = complex model, low = smooth model)

````python
import numpy as np
from sklearn.svm import SVR

X = np.array([[1], [2], [3], [4]])
y = np.array([2, 4, 6, 8])

model = SVR(kernel='linear')
model.fit(X, y)

print(model.predict(X))
````

### **Common Methods (ALL models)**

```bash
model.fit(X_train, y_train)
model.predict(X_test)
model.score(X_test, y_test)
```

**model.score()** : score() is used to evaluate how well your model is performing on given data.

1. For Classification Models it  Returns accuracy
    * **Example:** 80 correct out of 100 → score = 0.8
2. For Regression Models it Returns R² Score <br>
**Range:**
* 1 → perfect prediction
* 0 → average
* < 0 → very bad model
